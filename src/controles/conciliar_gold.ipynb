Celda 1: los parámetros (en qué esquemas buscar):


```python
# Conciliación: gold se calcula desde silver, así que el petróleo total tiene que ser el mismo.
# Los esquemas llegan como parámetros de la tarea: en dev y en prod son distintos.
dbutils.widgets.text("silver", "")
dbutils.widgets.text("gold", "")
silver = dbutils.widgets.get("silver")
gold = dbutils.widgets.get("gold")
```

Celda 2: el control:



```python
total_silver = spark.sql(f"SELECT round(sum(prod_pet), 2) AS t FROM {silver}.produccion").first()["t"]
total_gold = spark.sql(f"SELECT round(sum(prod_pet), 2) AS t FROM {gold}.produccion_pozo_mes").first()["t"]
print(f"Petróleo en silver: {total_silver} m3 | en gold: {total_gold} m3")

# Si no cuadra, la tarea falla: el job queda en rojo y llega el mail.
if total_silver != total_gold:
    raise ValueError(f"Gold no cuadra con silver: {total_gold} m3 contra {total_silver} m3")
```
